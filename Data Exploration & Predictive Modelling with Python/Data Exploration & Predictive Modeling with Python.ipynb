```python
import pandas as pd

dataset = pd.read_csv("data.csv.xls")
```

```python
#Columns are called variables in machine learning. 
#step 1: check for relationship between the variables or columns
#step 2: identify the independent and dependent variables
#step 3: call the algorithm
#step 4: create a linear regression model
#step 5: fit/train the model
#step 6: predict on x values
```

```python
#step 2:
#identify the independent and dependent variables
#x is mostly denoted for independent,
#has more that one brackets(double),
#means it can have more columns,
#must be two dimensional

#y is mostly denited for dependent
#i.e y depends on x
```

```python
import seaborn as sns
sns.scatterplot(x="Height", y="Weight", data=dataset)
```

```python
x= dataset[["Height"]].values
y= dataset["Weight"].values
```

```python
#to check x dimension
x.ndim
```

```python
#to check y dimension
y.ndim
```

```python
#step 3:
#call the algorithm
import sklearn
#sklearn is also sci_kit learn. it is a library for machine learning
```

```python
from sklearn.linear_model import LinearRegression
```

```python
#step 4:
#create a linear regression model
model = LinearRegression
```

```python
#step 5:
#fit/train the model
model = LinearRegression().fit(x,y)
```

```python
#step 6:
#predict on x values
model.predict(x).round(2)
```

```python
x
```

```python
y
```

```python
#step 7:
#checking the accuracy of a linear regression
from sklearn.metrics import r2_score
```

```python
prediction = model.predict(x).round(2)
```

```python
r2_score(y,prediction)
```

```python
#predict the weight of a woman with height 1.89
model.predict([[1.89]]).round(2)
```

```python
model.predict([[1.89]]).round(0)
```

```python
#step 8:
#plotting the regression line(ci means confidence interval)
sns.lmplot(x="Height", y="Weight", data=dataset, line_kws={"color": "red"}, ci=None)
```
