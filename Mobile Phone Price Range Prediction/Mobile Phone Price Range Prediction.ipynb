```python
#step 1
x = mpp.drop(["Price ($)"], axis=1)
y = mpp["Price ($)"]
```

```python
#correlation
mpp.corr()["Price ($)"]
```

```python
#step 2:
#split x and y
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.6, random_state=0)
```

```python
#feature scaling
#feature scaling is important when we have large values and small values in the columns

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.fit_transform(x_test)
```

```python
#feature scaling
#feature scaling is important when we have large values and small values in the columns

from sklearn.preprocessing import StandardScaler

x_train = StandardScaler().fit_transform(x_train)
x_test = StandardScaler().fit_transform(x_test)
```

```python
#step 3:
#call the algorithm
from sklearn.svm import SVR

model = SVR(kernel="poly")
```

```python
#step 4:
#train the model
model = model.fit(x_train, y_train)
```

```python
#step 5:
#make predictions on x_test values
model.predict(x_test).round(0)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test, model.predict(x_test))*100
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_test, model.predict(x_test)).round()
```
