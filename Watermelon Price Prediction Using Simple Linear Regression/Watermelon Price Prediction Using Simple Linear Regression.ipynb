```python
import pandas as pd
watermelon = pd.read_csv("Watermelon price prediction.csv")
```

```python
watermelon
```

```python
watermelon.dtypes
```

```python
#to show if there is a linear relationship between the price and weight
import seaborn as sns
sns.scatterplot(x="Weight", y="Price", data=watermelon)
```

```python
#step 1:
#identify the x and y
x= watermelon.drop("Price", axis=1)
```

```python
#OR
x = watermelon[["Weight"]]
y = watermelon["Price"]
```

```python
x
```

```python
y
```

```python
#step 2:
#split x and y
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.5, random_state=0)
```

```python
x_train, y_train
```

```python
#step 3:
#call the algorithm and import the model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
#step 4:
#train the model
model = model.fit(x_train,y_train)
```

```python
#model is ready for predictions
#we use x_test to assess the model, i.e work to do
```

```python
#step 5:
#make predictions on x_test values
prediction = model.predict(x_test).round(0)
```

```python
x_test, y_test
```

```python
#step 6:
#predict the price of a weight of a watermelon not in the table
#predict the price of watermelon with

model.predict([[11]]).round()
```

```python
#predicted price for weight 11 is 462
```

```python
#step 7:
#evaluate the model
#check the accuracy
from sklearn.metrics import r2_score
r2_score(y_test, prediction)*100
```

```python
#mean squared erroe

from sklearn.metrics import mean_squared_error
mean_squared_error(y_test, prediction)
```
