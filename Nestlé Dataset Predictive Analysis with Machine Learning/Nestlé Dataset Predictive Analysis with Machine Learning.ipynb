```python
stock_price = stock_price.fillna(0)
```

```python
#converting a column to date
stock_price["Date"] = pd.to_datetime(stock_price["Date"])
```

```python
#extracting the year "Y" from the entire date
stock_price["Date"].dt.to_period("Y")
```

```python
stock_price.insert(1, "Year", stock_price["Date"].dt.to_period("Y"))
```

```python
stock_price
```

```python
stock_price.dtypes
```

```python
stock_price["Year"] = stock_price["Year"].astype(str).astype(int)
```

```python
stock_price.dtypes
```

```python
#Building the ML model
#step 1:
#Identify the independent(features) and dependent(target) columns

x = stock_price.drop(["Date", "Close Price"], axis=1)
y = stock_price["Close Price"]
```

```python
x
```

```python
y
```

```python
#step 2:
#split x and y
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.6, random_state=0)
```

```python
#step 3:
#call the algorithm

from sklearn.linear_model import LinearRegression
model =  LinearRegression()
```

```python
#step 4:
#train the model 
model = model.fit(x_train, y_train)
```

```python
#step 5:
#make predictions on the x_test values

prediction = model.predict(x_test)
```

```python
prediction.round(0)
```

```python
#step 6:
#check the accuracy
from sklearn.metrics import r2_score
r2_score(y_test, prediction)*100
```
