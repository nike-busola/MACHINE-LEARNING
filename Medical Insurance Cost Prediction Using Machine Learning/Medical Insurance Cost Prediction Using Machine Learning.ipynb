```python
import pandas as pd
insurance = pd.read_csv("Insurance data.csv")
```

```python
insurance
```

```python
insurance.dtypes
```

```python
#to show if there is a linear relationship between the price and weight
import seaborn as sns
sns.scatterplot(x="Age", y="Premium", data=insurance)
```

```python
#step 1:
#identify the x and y
x = insurance.drop("Premium", axis=1)
y = insurance["Premium"]
```

```python
x
```

```python
y
```

```python
#step 2:
#split x and y
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.5, random_state=0)
```

```python
x_train, y_train
```

```python
#step 3:
#call the algorithm and import the model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
#step 4:
#train the model
model = model.fit(x_train,y_train)
```

```python
#step 5:
#make predictions on x_test values
prediction = model.predict(x_test).round(0)
```

```python
x_test, y_test
```

```python
#step 6:
#evaluate the model
#check the accuracy
from sklearn.metrics import r2_score
r2_score(y_test, prediction)*100
```

```python
#mean squared error

from sklearn.metrics import mean_squared_error
mean_squared_error(y_test, prediction)
```
